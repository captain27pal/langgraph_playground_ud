```python
# Install the missing package in the active notebook kernel
# %pip install langchain-openai

# Use the already-defined API key from the earlier cell
from langchain_openai import ChatOpenAI
from langchain_anthropic import ChatAnthropic

llm_openai = ChatOpenAI(
        model="gpt-4o-mini",
        api_key=open_ai_api_key,
        temperature=0
)
llm_anthropic = ChatAnthropic(
    model="claude-sonnet-4-6",
    api_key=anthropic_api_key,
    temperature=0
)
```

```python
from langchain_core.messages import HumanMessage
msg = HumanMessage(content="Hello, how are you?, whats my name", name = "utpal")
llm_response_openai = llm_openai.invoke([msg])
llm_response_anthropic = llm_anthropic.invoke([msg])
print(llm_response_openai.content, "\n","****************\n")
print(llm_response_anthropic.content)
```

```python
# %pip install langchain_community
from langchain_community.tools.tavily_search import TavilySearchResults
tavily_search = TavilySearchResults(max_results = 3)
search_docs = tavily_search.invoke("What is langgraph?")
```
